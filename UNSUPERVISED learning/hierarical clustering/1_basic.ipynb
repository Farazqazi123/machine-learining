```python
from sklearn.preprocessing import  StandardScaler
scaler=StandardScaler()

x_scaled=scaler.fit_transform(df)
```

```python
x_scaled
```

```python
x_scaled.shape
```

```python
from sklearn.decomposition import PCA

pca=PCA(n_components=2)
```

```python
scaled_pca=pca.fit_transform(x_scaled)
```

```python
plt.scatter(scaled_pca[:,0],scaled_pca[:,1])
```

```python
import scipy.cluster.hierarchy as sc

plt.Figure(figsize=(20,7))
plt.title("dendogram")

sc.dendrogram(sc.linkage(scaled_pca,method="ward"))
plt.xlabel("sample index")
plt.ylabel("eucladien distance")
```

```python
from sklearn.cluster import AgglomerativeClustering

model= AgglomerativeClustering(n_clusters=2,linkage="ward")

labels=model.fit_predict(scaled_pca)

```

```python
labels
```

```python
plt.scatter(scaled_pca[:,0],scaled_pca[:,1], c=labels)
plt.plot
```

```python

# performance validation we use seloid score

from sklearn.metrics import silhouette_score
```

```python
silhouette_coefficients=[]
for k in range(2,11):
    aglo=AgglomerativeClustering(n_clusters=k, linkage="ward")
    aglo.fit(x_scaled)
    score=silhouette_score(x_scaled,aglo.labels_)
    silhouette_coefficients.append(score)
```
